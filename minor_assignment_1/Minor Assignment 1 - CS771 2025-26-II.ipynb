```python
import numpy as np
from sklearn.svm import LinearSVC

import genSyntheticData as gsd
import plotData as pd

from matplotlib import pyplot as plt
import time
```

```python
# The genChallengeData function takes two arguments
# level should be a non-negative number strictly greater than zero
# n should be a non-negative integer as it decides the number of points
myLevel = 100
XPos, yPos, XNeg, yNeg = gsd.genChallengeData( level = myLevel, n = 50 )

X = np.vstack( (XPos, XNeg) )
y = np.concatenate( (yPos, yNeg) )

fig = pd.getFigure( 2 * np.sqrt( 2 ) * ( 1 + 1 / ( 2 * myLevel ) ) + 2, 2 * np.sqrt( 2 ) + 2 )
pd.plot2D( XPos, fig, color = 'm', marker = '+' )
pd.plot2D( XNeg, fig, color = 'y', marker = 'o' )
plt.show()
```

```python
myPenalty = "l2"
myC = 0.1
myLoss = "squared_hinge"
myMaxIter = 100
myTol = 1e-4
clf = LinearSVC( penalty = myPenalty, loss = myLoss, C = myC, tol = myTol, max_iter = myMaxIter )

n_runs = 5
t_train = 0
for t in range( n_runs ):
    tic = time.perf_counter()
    clf.fit( X, y )
    toc = time.perf_counter()
    t_train += toc - tic
t_train /= n_runs

acc = np.average( y == clf.predict( X ) )

w = clf.coef_.reshape(-1)
b = clf.intercept_

mySVM = lambda X: X.dot(w) + b

fig3 = pd.getFigure( 2 * np.sqrt( 2 ) * ( 1 + 1 / ( 2 * myLevel ) ) + 2, 2 * np.sqrt( 2 ) + 2 )
pd.shade2D( mySVM, fig3, mode = 'batch', xlim = np.max( np.abs( X[ :, 0 ] ) ) * 1.1, ylim = np.max( np.abs( X[ :, 1 ] ) ) * 1.1 )
pd.plot2D( XPos, fig3, color = 'm', marker = '+' )
pd.plot2D( XNeg, fig3, color = 'y', marker = 'o' )
plt.title( f"On challenge level {myLevel}, using {myPenalty} penalty, {myLoss} loss,\nC = {myC}, max_iter {myMaxIter}, and {myTol} tol yeilded\n{acc:.2f} accuracy in {t_train:.4f} sec train time" )
plt.show()
```

```python
l = 0.3
penalty = 'l2'
loss_fn = 'squared_hinge'
Xpos, ypos, Xneg, yneg = gsd.genChallengeData(level = l, n = 1000)
X = np.vstack((Xpos, Xneg))
y = np.concatenate((ypos, yneg))

fig = pd.getFigure( 2 * np.sqrt( 2 ) * ( 1 + 1 / ( 2 * l ) ) + 2, 2 * np.sqrt( 2 ) + 2 )
pd.plot2D( Xpos, fig, color = 'm', marker = '+' )
pd.plot2D( Xneg, fig, color = 'y', marker = 'o' )
plt.show()
```

```python
tol_vals = [1e-2,1e-1,1,2.5,5]
C_vals = [1e-3,1e-2,1e-1,1]
max_iter_vals = [1,2,3,4,5,10]

max_acc = 0
best_features = [0, 0, 0]
best_clf = LinearSVC()
best_ttime = 0;

for t in tol_vals:
    for c in C_vals:
        for m_iter in max_iter_vals:
            clf = LinearSVC(penalty='l2', loss='squared_hinge', tol=t, C=c, max_iter=m_iter)
            tic = time.perf_counter()
            clf.fit(X, y)
            toc = time.perf_counter()
            
            accuracy = np.average( y==clf.predict(X))
            print(f"C: {c}, max_iterations: {m_iter}, tolerance: {t} -> Accuracy: {accuracy*100}%")
            
            if(max_acc < accuracy):
                max_acc = accuracy
                best_features = [t, c, m_iter]
                best_clf = clf
                best_ttime = toc-tic
            if(max_acc == accuracy):
                best_features[0] = max(best_features[0], t)
                best_features[1] = min(best_features[1], c)
                best_features[2] = min(best_features[2], m_iter)
                best_ttime = min(best_ttime, toc-tic)
                
print("best accuracy: ", max_acc)
print("features: ", best_features)

w = best_clf.coef_.reshape(-1)
b = best_clf.intercept_

svm_clf = lambda X: X.dot(w) + b

fig3 = pd.getFigure( 2 * np.sqrt( 2 ) * ( 1 + 1 / ( 2 * l ) ) + 2, 2 * np.sqrt( 2 ) + 2 )
pd.shade2D( svm_clf, fig3, mode = 'batch', xlim = np.max( np.abs( X[ :, 0 ] ) ) * 1.1, ylim = np.max( np.abs( X[ :, 1 ] ) ) * 1.1 )
pd.plot2D( Xpos, fig3, color = 'm', marker = '+' )
pd.plot2D( Xneg, fig3, color = 'y', marker = 'o' )
plt.title( f"On challenge level {l}, using {penalty} penalty, {loss_fn} loss,\nC = {best_features[1]}, max_iter {best_features[2]}, and {best_features[0]} tol yeilded\n{max_acc:.2f} accuracy in {best_ttime:.4f} sec train time" )
plt.savefig('Q2_plot.png', bbox_inches='tight')
plt.show()            
```

```python
n_values = [10, 50, 100, 500, 1000, 1500, 2000, 2500, 3000, 3500, 4000, 4500, 5000]
l = 100
results = []
y_vals = []

for n in n_values:
    # 1. Generate data
    Xpos, ypos, Xneg, yneg = gsd.genChallengeData(level=l, n=n)
    X = np.vstack((Xpos, Xneg))
    y = np.concatenate((ypos, yneg))
    
    # 2. Initialize classifier (use enough max_iter for high challenge levels)
    clf = LinearSVC(penalty = 'l2', loss='squared_hinge', C=10000, max_iter=10000, tol=1e-7)
    
    # 3. Measure training time
    t_time = 0
    for i in range(0,20):
        start = time.time()
        clf.fit(X, y)
        end = time.time()
        t_time += end-start
    
    # 4. Record accuracy and time
    accuracy = clf.score(X, y)
    train_time = (t_time/20) * 1000 # convert to microseconds
    results.append((n, accuracy, train_time))
    y_vals.append(train_time)

print("RESULTS: ")
print(results[0])
print(results[1])
print(results[2])
print(results[3])
print(results[4])
print(results[5])
print(results[6])
print(results[7])
print(results[8])
print(results[9])
print(results[10])
print(results[11])
print(results[12])
```

```python
plt.plot(n_values, y_vals, label='Training Time', marker='o') # 'o' adds dots to the points
plt.xlabel('Number of Points (n)')
plt.ylabel('Training Time (milli seconds)')
plt.title('Training Time vs. Number of Points')
plt.legend() # Now it has a label to show
plt.grid(True, linestyle='--', alpha=0.6) # Optional: makes it easier to read values
plt.savefig('Q5_plot.png')
plt.show()
```
